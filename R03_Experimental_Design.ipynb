### ✅ 1. Bonferroni Correction – Overall Alpha vs Number of Tests

* __Bonferroni correction adjusts the per-test significance level__ to:  __$𝛼_{𝐵𝑜𝑛𝑓𝑒𝑟𝑟𝑜𝑛𝑖}=\frac{𝛼}{𝑘}$__

* The formula: $𝐹𝑊𝐸𝑅=1−(1−𝛼_{𝐵𝑜𝑛𝑓𝑒𝑟𝑟𝑜𝑛𝑖} )^𝑘$

represents the __actual overall probability of making at least one Type I error__ when each test uses the __Bonferroni-adjusted alpha.__

This shows whether the Bonferroni correction successfully controls the overall error rate.

This code demonstrates how the __Bonferroni correction__ controls the __overall Type I error rate (α)__ when performing multiple hypothesis tests.

When multiple tests are conducted simultaneously, the probability of making at least __one false positive increases.__ The __Bonferroni method__ adjusts the significance level to keep the __overall Type I error__ at the desired level (here, __α = 0.05__).

```r
# Step 1: Define significance level
alpha <- 0.05   # Desired overall Type I error rate

# Step 2: Define number of tests
k <- 4:50       # Number of simultaneous tests

# Step 3: Compute overall alpha
# Formula: 1 - (1 - alpha/k)^k
overall_alpha <- 1 - (1 - alpha/k)^k

# Step 4: Plot the curve
plot(k, overall_alpha,
     type = "l",
     lwd = 2,
     col = "blue",
     xlab = "k (Number of Tests)",
     ylab = "Overall value for alpha(Type I Error Rate)",
     main = "Bonferroni Correction Effect")

# Add reference line at alpha = 0.05
abline(h = alpha, lty = 2, col = "red")

# Add legend
legend("topright",
       legend = c("Overall alpha with Bonferroni", "Nominal alpha = 0.05"),
       col = c("blue", "red"),
       lwd = c(2, 1),
       lty = c(1, 2))
```

### ➡️ Interpretation
➡️ __Step 1:__ Define significance level: 

     alpha <- 0.05
* This is the desired overall Type I error rate.
* It means we want only a 5% chance of making at least one false positive error overall.

➡️ __Step 2:__ Define number of tests: 
         
      k <- 4:50
* The code considers different numbers of hypothesis tests, from 4 to 50 tests.
* As the number of tests increases, the risk of false positives increases without correction.


➡️ __Step 3:__ Compute overall alpha after Bonferroni correction:

     overall_alpha <- 1 - (1 - alpha/k)^k

➡️ __Step 4:__ Plot interpretation

The plot shows:

* __X-axis:__ Number of tests (k)
* __Y-axis:__ Overall Type I error rate
* __Blue line:__ Actual overall alpha after Bonferroni correction
* __Red dashed line:__ Desired alpha level (0.05)

📌 __1. The blue curve stays close to 0.05__

This means:

* Bonferroni correction successfully keeps the overall Type I error near the desired level (0.05)
* Even when the number of tests increases from 4 to 50

📌 __2. The overall alpha slightly decreases as k increases__

This indicates:

* Bonferroni correction becomes more conservative when the number of tests increases
* It reduces the chance of false positives even further than required

📌 __3. Bonferroni prevents inflation of false positives__

Without correction, overall alpha would increase rapidly with more tests.

__For example:__

* Without correction:

      20 tests → overall error ≈ 64%

* With Bonferroni correction:

      overall error remains ≈ 5%


#### 📌 Practical Meaning

Bonferroni correction:

✅ Prevents false positives

✅ Maintains overall error rate

❌ May reduce power (harder to detect true effects)

 __👉 Bonferroni correction keeps the overall Type I error rate near 0.05 even when many tests are performed, but it becomes more conservative as the number of tests increases.__

* This function identifies influential observations in the regression model.

* The plot contains 4 panels, each indexed by observation number.

__🔹1. Cook’s Distance__

Measures overall influence of each observation.

__Interpretation:__

* Large Cook’s distance → observation strongly influences regression results

* These points may change regression coefficients significantly

Rule of thumb:
$Cook′s\; Distance > \frac{4}{n}$   → considered influential

__🔹2. Studentized Residuals__

Measures how far observed value is from predicted value.

__Interpretation:__

* Large positive value → observation higher than expected
* Large negative value → observation lower than expected
* Absolute value > 2 or 3 → possible outlier

__🔹3. Bonferroni p-value__

Tests whether an observation is a significant outlier after multiple testing correction.

__Interpretation:__

* Small Bonferroni p-value (< 0.05) → strong evidence of outlier*
* Large value → observation is not a significant outlier

Bonferroni correction prevents false positives due to multiple testing.

__🔹4. Hat Values (Leverage)__

Measures how extreme predictor values are.

__Interpretation:__

* High leverage → observation has unusual predictor values

* These points can strongly affect regression slope

Rule of thumb: $Hat\; value >\frac{2p}{n}$ 

where p = number of parameters

### ✅ 3. OutlierTest

```r
# Load required libraries
library(car)
library(carData)

# Fit a linear regression model using the Duncan dataset
Model <- lm(prestige ~ income + education, data = Duncan)

# Perform an outlier test on the model
outlierTest(Model)

# Simple numeric calculation
0.0031772 * 45
```

### ➡️ Interpretation

📌 1. Fits a linear regression model: $Prestige=β_0+β_1​(Income)+β_2​(Education)+ϵ$  using the Duncan dataset.

📌 2. This function: 

     outlierTest(Model)

* Computes Studentized residuals
* Performs a Bonferroni-adjusted test to detect significant outliers
* Corrects for multiple testing across all observations

1️⃣ __Largest Studentized Residual__

       Occupation: minister
       Studentized residual = 3.13

__Interpretation:__

* This value is relatively large (greater than 3).
* It suggests that "minister" deviates noticeably from the regression line.
* It may appear to be a potential outlier.

2️⃣ __Unadjusted p-value__

     p = 0.0031772

__Without correction:__
* This would be considered statistically significant (p < 0.05).
* So initially, it looks like an outlier.

3️⃣ Bonferroni Correction

* To calculated: $0.0031772×45=0.14297$  There are 45 observations in the dataset.

* Bonferroni adjustment:  $p_{adjusted}=p×n= 0.0031772×45=0.14297$

* Corrected p-value = 0.14297

#### 📌 Final Statistical Decision

Since: $0.14297>0.05$

👉 The observation is NOT statistically significant after Bonferroni correction.

This is why the output states:__No Studentized residuals with Bonferroni p < 0.05__


__👉 Conclusion: There are no statistically significant outliers in the regression model after adjusting for multiple testing.__

__The observation "minister" shows the largest deviation from the model, but it does not significantly influence the model at the 5% significance level once Bonferroni correction is applied__

### ✅ 4. Post-hoc power calculation using power.t.test in R
Step 1: Define parameters
* n          = sample size per group (here 15)
* delta      = true difference in means (effect size)
* sd         = assumed standard deviation (default = 1 if not specified)
* sig.level  = significance level (alpha = 0.05)
* alternative= type of test ("two.sided", "one.sided")

```r
# Step 2: Run the power calculation
result <- power.t.test(n = 15,
                       delta = 1,
                       sig.level = 0.05,
                       alternative = "two.sided")

# Step 3: Display the result
print(result)
```

### ➡️ Interpretation

👉 Our experiment with 15 subjects per group, Effect size = 1 at significance level 0.05 with 75.3% power
Meaning the study had a reasonably good but slightly insufficient ability to detect the true difference

### ✅ 5. A priori power calculation for a two-sample t-test

Step 1: Load R (no extra packages needed, power.t.test is built-in)

Step 2: Define parameters
* power      = desired statistical power (here 0.8 = 80%)
* delta      = expected difference in means (effect size)
* sd         = assumed standard deviation (default = 1 if not specified)
* sig.level  = significance level (alpha = 0.05)
* alternative= type of test ("two.sided", "one.sided")

```r
# Step 3: Run the power calculation
result <- power.t.test(power = 0.8,
                       delta = 1,
                       sig.level = 0.05,
                       alternative = "two.sided")

# Step 4: Display the result
print(result)
```

### ➡️ Interpretation

👉 To detect a true difference of 1 (with SD = 1) at significance level 0.05 with 80% power, we need at least: 17 observation per group.


### ✅ 6. Power curve using power.t.test (theoretical)
Parameters:
* delta = expected mean difference (18 - 17 = 1)
* sd    = assumed standard deviation (2)
* sig.level = 0.05
* power.t.test requires either n or power, so we loop over n

```r
# -----------------------------
# PARAMETERS
# -----------------------------
alpha  <- 0.05        # significance level
delta  <- 1           # true mean difference (18 - 17)
sd     <- 2           # standard deviation
NN     <- 3:200       # sample sizes
nsim   <- 1000        # number of simulations

# -----------------------------
# 1️. Analytical Power (power.t.test)
# -----------------------------
Power1 <- numeric(length(NN))

for(j in 1:length(NN)){
  Power1[j] <- power.t.test(n = NN[j],
                             delta = delta,
                             sd = sd,
                             sig.level = alpha,
                             type = "two.sample",
                             alternative = "two.sided")$power
}


# Plot the power curve
plot(NN, Power1,
     type = "l",
     lwd = 2,
     xlab = "NN",
     ylab = "Power",
     main = "Power Curve: Analytical Power (power.t.test)")

# Add horizontal line at 0.8 power
abline(h = 0.8, lty = 2)

# Find required sample size for 80% power
n_required <- min(NN[Power1 >= 0.8])

# Add vertical line at required n
abline(v = n_required, lty = 2)

# Add text label near bottom
text(n_required, 0.05, labels = n_required)
```

### ➡️ Interpretation

Interpretation of the Graph
* Small n → low power
* Around n ≈ 64 → Power ≈ 0.8
* Large n → power approaches 1

The dashed horizontal and vertical lines in your figure correspond to:
__Power=0.8 and n≈64__


### ✅ 7. Power Calculations Using Simulation
Estimate power of a t-test by simulating samples

```r
# Step 1: Set seed for reproducibility
set.seed(658)

# Step 2: Simulate sampling for two groups
Males   <- round(rnorm(n = 15, mean = 18, sd = 2), 2)
Females <- round(rnorm(n = 15, mean = 17, sd = 2), 2)

# Step 3: Perform one t-test on simulated samples
t_result <- t.test(Males, Females, var.equal = TRUE)

# Step 4: Display full t-test result
print(t_result)

# Step 5: Extract and display only the p-value
p_value <- t_result$p.value
print(p_value)
```

__We fail to reject $𝐻_0$__

👉 This is a __Type II error__ in this specific simulated experiment.

__📌 Confidence Interval: (−0.356, 3.052)__

### ✅ 8. Power Calculation via 10,000 Simulated t-tests
__👉 Estimate power for $n = 20$, effect size $= 1\; (18 - 17)$, SD $= 2$__

```r
# Step 1: Initialize storage for p-values
pval <- numeric(10000)

# Step 2: Run 10,000 simulations
for (i in 1:10000) {
  # Generate samples for two groups
  A <- rnorm(n = 20, mean = 18, sd = 2)
  B <- rnorm(n = 20, mean = 17, sd = 2)

  # Perform two-sample t-test and store p-value
  pval[i] <- t.test(A, B, var.equal = TRUE)$p.value
}

# Step 3: Calculate empirical power
# Proportion of p-values less than alpha = 0.05
empirical_power <- sum(pval < 0.05) / 10000

# Step 4: Display result
print(empirical_power)
```

### ➡️ Interpretation

$Power ≈ 0.3419\; (34.2$%)

__Meaning:__ P(Reject $𝐻_0$ ∣ True difference = 1) = $34$%

So $β = 1 − 0.3419 = 0.6581$

__👉 There is about 66% chance of missing the true effect.__

### ✅ 10. Simulation Power (Nested Loops)
__Nested loops:__
* First loop for 198 different sample sizes.
* Second loop, nested in first loop, to create 1000 t-tests.

```r
# Simulation Power (Nested Loops)

# Sample sizes from 3 to 200 (198 values)
NN <- 3:200
Power2 <- numeric(length(NN))            # Placeholder for power values
pval   <- numeric(1000)                  # Placeholder for p-values from 1000 simulations
set.seed(123)                            # For reproducibility

for(j in 1:length(NN)){                     # Loop over different sample sizes
  for(i in 1:1000){                         # Inner loop: 1000 simulated experiments
    # Generate samples
    A <- rnorm(NN[j], mean = 18, sd = 2)
    B <- rnorm(NN[j], mean = 17, sd = 2)
    pval[i] <- t.test(A, B, var.equal = TRUE)$p.value     # Perform two-sample t-test
  }
  Power2[j] <- mean(pval < alpha)                         # Estimate power
}

# Plot the simulated power curve
plot(NN, Power2,
     type = "l",
     pch = 1,
     lwd = 2,
     ylim = c(0,1),
     xlab = "Sample Size (per group)",
     ylab = "Power",
     main = "Power Curve (Simulation)")

# Add 80% power reference line
abline(h = 0.8, lty = 2)
```

### ➡️ Interpretation

__The Curve Shows__

👉 Small n → low power

👉 As n increases → power increases

👉 Curve approaches 1

👉 Around n ≈ 64 per group → power ≈ 0.8 (approximately)


### ✅ 11. Combined Power Curve (Analytical + Simulation)
__📌 To generate the combined power curve using:__

* Analytical method → power.t.test()  (theoretical power)

* Simulation method → Nested loops (1000 t-tests per sample size) (empirical power)

__📌 This matches your setup:__

* Two-sample t-test

* Mean1 = 18

* Mean2 = 17

* sd = 2

* α = 0.05

* Two-sided test

* Sample sizes: 3 to 200

* 1000 simulations per sample size

```r
# -----------------------------
# PARAMETERS
# -----------------------------
alpha  <- 0.05        # significance level
delta  <- 1           # true mean difference (18 - 17)
sd     <- 2           # standard deviation
NN     <- 3:200       # sample sizes
nsim   <- 1000        # number of simulations

# -----------------------------
# 1️. Analytical Power (power.t.test)
# -----------------------------
Power1 <- numeric(length(NN))

for(j in 1:length(NN)){
  Power1[j] <- power.t.test(n = NN[j],
                             delta = delta,
                             sd = sd,
                             sig.level = alpha,
                             type = "two.sample",
                             alternative = "two.sided")$power
}

# -----------------------------
# 2️. Simulation Power (Nested Loops)
# -----------------------------
Power2 <- numeric(length(NN))
pval   <- numeric(nsim)

set.seed(123)   # for reproducibility

for(j in 1:length(NN)){

  for(i in 1:nsim){
    A <- rnorm(NN[j], mean = 18, sd = sd)
    B <- rnorm(NN[j], mean = 17, sd = sd)

    pval[i] <- t.test(A, B, var.equal = TRUE)$p.value
  }

  Power2[j] <- mean(pval < alpha)
}

# -----------------------------
# 3️. Plot Combined Power Curve
# -----------------------------
plot(NN, Power2,
     type = "p",
     pch = 1,
     ylim = c(0,1),
     xlab = "Sample Size (n per group)",
     ylab = "Power",
     main = "Power Curve: Analytical vs Simulation")

lines(NN, Power1, col = "blue", lwd = 2)

# Add 80% reference line
abline(h = 0.8, lty = 2)

# Find sample size achieving 80% power (analytical)
n80 <- ceiling(power.t.test(power = 0.8,
                            delta = delta,
                            sd = sd,
                            sig.level = alpha,
                            type = "two.sample",
                            alternative = "two.sided")$n)

abline(v = n80, lty = 2)
text(n80, 0.05, labels = n80, pos = 3)

legend("bottomright",
       legend = c("Simulation", "Analytical (power.t.test)"),
       col = c("black", "blue"),
       pch = c(1, NA),
       lty = c(NA, 1),
       lwd = c(NA, 2))
```

### ➡️ Interpretation

__📌 The graph Shows__
* __Black circles__ → Simulated power (1000 t-tests per sample size)
* __Blue line__ → Exact theoretical power from power.t.test()
* __Horizontal dashed line__ → 80% power
* __Vertical dashed line__ → Required sample size per group 

__📌 Interpretation__
* The two curves should almost overlap (as in your image).
* Small differences occur due to Monte Carlo simulation error.
* The required n per group for 80% power is around 64 (as seen in figure).


### ✅ 12. Power Curve - Effect Size Given n (Sensitivity Analysis)
__Estimate power for different effect sizes at fixed sample size n = 20__


```r
# -----------------------------
# PARAMETERS
# -----------------------------
# alpha  <- 0.05
# sd     <- 2
# n      <- 20          # fixed sample size per group
#nsim   <- 1000
# -----------------------------

# Step 1: Adjust plot margins (optional, for later plotting)
par(mar = c(2, 2, 1, 2))

# Step 2: Define placeholders
Power3 <- numeric(39)   # Store power estimates for 39 effect sizes
pval   <- numeric(1000) # Store p-values for 1000 simulations per effect size

# Step 3: Define effect sizes
# Baseline mean = 17, effect sizes range from 17.2 to 21.0 in steps of 0.1
Eff_Siz <- 17 + seq(0.2, 4, 0.1)

# Step 4: Nested loops
# Outer loop: iterate over 39 effect sizes
# Inner loop: run 1000 t-tests for each effect size
for (j in 1:39) {
  for (i in 1:1000) {
    # Generate samples
    A <- rnorm(n = 20, mean = Eff_Siz[j], sd = 2)  # Group A with varying mean
    B <- rnorm(n = 20, mean = 17, sd = 2)          # Group B fixed mean

    # Perform t-test and store p-value
    pval[i] <- t.test(A, B, var.equal = TRUE)$p.value
  }

  # Calculate empirical power for this effect size
  Power3[j] <- sum(pval < 0.05) / 1000
}

# Step 5: Plot the sensitivity analysis (power vs effect size)
plot(Eff_Siz, Power3, pch = 1, lwd = 2, col = "blue",
     xlab = "Effect Size (Difference in Means)",
     ylab = "Power",
     main = "Power Curve: Sensitivity Analysis (n = 20)")

# Add 80% reference line
abline(h = 0.8, lty = 2)

# Find effect size giving ~80% power
eff80 <- Eff_Siz[min(which(Power3 >= 0.8))]

abline(v = eff80, lty = 2)
text(eff80, 0.05, round(eff80,1), pos = 3)
```

### ➡️ Interpretation

__📌 The graph Shows__
* __X-axis__ → Effect Size (mean difference)
* __Y-axis__ → Power
* __Dots__ → Simulated power values
* __Horizontal dashed line__ → 80% power
* __Vertical dashed line__ → Effect size needed for 80% power (~18.9 in your image)

__📌 Interpretation__
* With __n = 20__ per group and __sd = 2__:
* __Small effect sizes__ → low power
* __Around mean ≈ 18.8__ → power ≈ 0.8
* __Large effect sizes__ → power → 1

__👉 This is called __Sensitivity Analysis__ because we fix sample size and examine what effect size we can detect.__
